```python
import osmnx as ox
import pandas as pd
import folium
import matplotlib
import numpy as np
import geopandas as gpd
from shapely.geometry import MultiPolygon, Polygon
```

### 1. Get 3 polygons 

random sample from Missoula data

```python
gdf = gpd.read_file("../../../maps/mso_features.gpk").to_crs(4326)  # WGS84 for mapping

poly_sample = gdf[gdf.geometry.type.isin(['Polygon'])].sample(3)

poly_sample.explore(tiles="CartoDB positron")
```

### 2. Save as polygons

```python
poly_sample.to_file("three_polygons.shp", driver="ESRI Shapefile")
poly_sample.to_file("three_polygons.geojson", driver="GeoJSON")
```

### 3. Save as MultiPolygon


collapse three Polygons into a single MultiPolygon object

```python
# generate shapely MultiPolygon object (no projection)
multipoly = MultiPolygon(poly_sample["geometry"].to_list())

# generate GeoDataFrame with one entry in WGS84
multi_gdf = gpd.GeoDataFrame({"id":[1]}, geometry=[multipoly], crs=poly_sample.crs)

# check: just one record?
assert(len(multi_gdf) == 1)
```

visual check: still 3 disjoint shapes

```python
multi_gdf.explore(tiles="CartoDB positron")
```

```python
multi_gdf.to_file("one_multipolygon.shp", driver="ESRI Shapefile")
multi_gdf.to_file("one_multipolygon.geojson", driver="GeoJSON")
```
